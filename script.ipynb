```python
import pandas as pd
#biblioteca para manipulação de dados
```

```python
#ler o arquivo Excel
df = pd.read_excel('tabela_vendas.xlsx')

df.info() #mostra informações sobre a tabela que foi lida como: colunas, quantidade de colunas e qual ordem, tipos de dados
```

```python
df['Produto'].value_counts()
```

```python
#formatando a tabela PRODUTO
df['Produto'] = df['Produto'].str.upper()

#lista todos os valores da tabela PRODUTO
df['Produto'].unique()
```

```python
#formatando a tabela PRODUTO removendo escritas incorretas e juntando dados

#substituindo iphones
df['Produto'] = df['Produto'].str.replace('IPHONE13', 'IPHONE 13')

#substituindo teclados redragon escritos errado
df['Produto'] = df['Produto'].str.replace('TECLADO MECÃ‚NICO REDRAGON', 'TECLADO MECÂNICO REDRAGON')
df['Produto'] = df['Produto'].str.replace('TECLADO MECÃ¢NICO REDRAGON', 'TECLADO MECÂNICO REDRAGON')

#listando produtos para conferir se a formatação funcionou
df['Produto'].unique()
```

```python
#tratando valores nulos da coluna QUANTIDADE

#df['Quantidade']: A coluna que vai receber o dado.

#df.groupby('Produto')['Quantidade']: Diz ao Python para olhar para a quantidade, mas separado por "pastas" de produtos.

#.transform(lambda x: x.fillna(x.mean())): Esse é o "cérebro". Para cada grupo (x), ele tenta preencher o nulo com a média daquele grupo específico.#

df['Quantidade'] = df.groupby('Produto')['Quantidade'].transform(lambda x: x.fillna(x.mean()))
```

```python
df['Quantidade'].isnull().sum()
```

```python
df['Produto'] = df['Produto'].str.replace('PLACA DE VÃ­DEO RTX 3060', 'PLACA DE VIDEO RTX 3060')

df['Produto'].unique()
```

```python
#transformando a coluna em inteiro após preencher os nulos com as medias.

df['Quantidade'] = df['Quantidade'].astype(int)

print(df['Quantidade'])
```

```python
df = df.reset_index(drop=True)
```

```python
#salvando ETL feito para a próxima fase de visualizações
df.to_csv('vendas_formatado.csv', index=False)
```
